## LINEAR TIME COMPLEXITY O(n)

```python
x = int(input("x ix "))
n = int(input("n ix "))
for i in range(n):
    if i == x:
        break;
    else:
        print(i);
    
```

## Constant Time Complexity O(1)

```python
k = 20000
l = 30000
#O(1)
for i in range(4):
    for i in range(3):
        print(k,l,i,end=" | ")
```

### O(n) + O(n^n) = O(n + n^n) ~ O(n^n)

```python
for i in range(n):# O(n^2)
        for i in range(n):
            print(i,end=" ")

for i in range(n):# O(n)
        print(i,end=" ")

```

## Logarithmic Time Complexity O(logn)

```python
i = 1
f = 100
# O(log2n)
while i<f:
    print(i,end=" ")
    i*=2
```

```python
i = 100
# O(log3n)
while i > 0:
    print(i,end=" ")
    i//=3
```

```python
for i in range(f):
    while i<f:
        print(i,end=" ")
        i*=2
```
